```python
### Remove duplicates from a CSV file and save the cleaned list to a new CSV file
### CSV files are assumed to have one word per line, with no header

import pandas as pd

# 1. Read the CSV file into a DataFrame
df = pd.read_csv('raw_words_bank.csv', header=None, names=['word'])

# 2. Remove the duplicates
# This keeps the first occurrence of each word and drops the rest
df_unique = df.drop_duplicates(subset=['word'])

# 3. Save the cleaned list to a new CSV file
# df_unique.to_csv('vocabulary_bank.csv', header=False, index=False)

# Optional: Print the results to see how many duplicates were removed
print(f"Original word count: {len(df)}")
print(f"Unique word count: {len(df_unique)}")
```

```python
df_unique['seq_len'] = df_unique['word'].str.len() *2 + 1 # assuming same length for the misspelled word in inference

print("Max seq_len:", df_unique['seq_len'].max())
print("99th percentile:", df_unique['seq_len'].quantile(0.99))
```
